```python
import sys
import numpy as np

# rows = []

# for line in sys.stdin:
#     if not line.strip():
#         continue

#     row = [
#         [int(channel) for channel in pixel.split(",")]
#         for pixel in line.split()
#     ]
#     rows.append(row)

# image = np.array(rows, dtype=np.uint8)

def read_csv(path):
    with open(path, "r", encoding="utf-8") as f:
        sys.stdin = f

        rows = []
        for line in sys.stdin:
            if not line.strip():
                continue

            row = [
                [int(channel) for channel in pixel.split(",")]
                for pixel in line.split()
            ]
            rows.append(row)

        image = np.array(rows, dtype=np.uint8)
        return image

test_img = read_csv("data.txt")
print("Shape:", test_img.shape)
```

```python
day_img = read_csv("day.txt")
print("Shape:", day_img.shape)
```

```python
night_img = read_csv("night.txt")
print("Shape:", night_img.shape)
```

```python
rgb = day_img.astype(np.float64) / 255.0
top = rgb[:max(1, int(rgb.shape[0] * 0.10))]

score = top.mean()
print("day" if score > 0.35 else "night")
```
